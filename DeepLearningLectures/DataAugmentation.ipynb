```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img, img_to_array
import requests
```

```python
datagen = ImageDataGenerator(
    rotation_range=20,
    width_shift_range=0.1,
    height_shift_range=0.1,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    vertical_flip=True,
    fill_mode='nearest'
)
```

```python
img_path = '/content/charizard.jpeg'
```

```python
plt.imshow(plt.imread(img_path))
```

```python
image_path = load_img(img_path)
image_array = img_to_array(image_path)
image_array = np.expand_dims(image_array, axis=0)
```

```python
augmented_images = []
for batch in datagen.flow(image_array, batch_size=1):
    augmented_images.append(batch[0].astype(np.uint8))
    if len(augmented_images) >= 10:
        break
```

```python
plt.figure(figsize=(16, 16))
for i, image in enumerate(augmented_images):
    plt.subplot(1, 10, i+1)
    plt.imshow(image)
    plt.axis('off')
plt.tight_layout()
```
